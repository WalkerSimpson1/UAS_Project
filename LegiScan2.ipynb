```python
import pandas as pd
import numpy as np
import zipfile as zf
import json
```

```python
with zf.ZipFile("legiscan_data/getDatasetRaw2026-09-06.zip") as raw:
    names = raw.namelist()

print(len(names))
print(names[:10])
    
```

```python
bill_names = []

for name in names:
        if "/bill/" in name:
            bill_names.append(name)

print(len(bill_names))
```

```python
with zf.ZipFile("legiscan_data/getDatasetRaw2026-09-06.zip") as raw:
    read = raw.read(bill_names[0])

structure = json.loads(read)

print(structure.keys())
```

```python
print(structure["bill"])
```

```python
All_Bills = []

with zf.ZipFile("legiscan_data/getDatasetRaw2026-09-06.zip") as raw:
    for name in bill_names:
        read = raw.read(name)
        structure = json.loads(read)
        All_Bills.append(structure["bill"])
```

```python
keywords = ["uas", "drone", "unmanned", "uav"]
```

```python
UAS_Bills = []

for bill in All_Bills:
    for word in keywords:
        if word in bill["title"].lower() or word in bill.get("description", "").lower():
            UAS_Bills.append(bill)
            break

print(len(UAS_Bills))
```

```python
print(UAS_Bills[0])
```

```python
for bill in UAS_Bills:
    print(f"Bill number: {bill['bill_number']}, Bill title: {bill['title']}")
```

```python
import sqlite3 as sql
from datetime import date
```

```python
connection = sql.connect("UAS_Tracker.db")
```

```python
sql_string = """    INSERT INTO legislation (item_name, 
                             identifier, 
                             source_url, 
                             source_platform, 
                             date_added, 
                             legiscan_raw_json, 
                             legiscan_change_hash, 
                             state, 
                             bill_id)
                             
                    VALUES (?, ?, ?, ?, ?, ?, ?, ?, ?)
                             
                             """
```

```python
cursor = connection.cursor() 

for bill in UAS_Bills:
    cursor.execute( sql_string, (
        bill["title"],
        bill["bill_number"],
        bill["state_link"],
        "LegiScan",
        date.today().isoformat(),
        json.dumps(bill),
        bill["change_hash"],
        bill["state"],
        bill["bill_id"]
        )
        )

connection.commit()
```
