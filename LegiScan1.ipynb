```python
import os
import time
import requests
import pandas as pd
from datetime import date
from dotenv import load_dotenv

load_dotenv() 
print ("Libraries imported")
```

```python
API_KEY = os.getenv("LEGI_API_KEY")
```

```python
url = "https://api.legiscan.com/"
## I need to get a access code from the getDatasetList url used to call the GetDataset url.
```

```python
params = {
    "key": API_KEY,
    "op": "getDatasetList",
    "state": "US"
}

response = requests.get(url, params=params)

print(response.status_code)

data = response.json()
print(data.keys())
```

```python
status = data["status"]
print(status)
```
